```python
df=df.drop(["sample_id"],axis=1)
```

```python
from sklearn.preprocessing import LabelEncoder
le_behav = LabelEncoder()
le_cir = LabelEncoder()

y_cir=le_cir.fit_transform(df["circuit"])
y_behav=le_behav.fit_transform(df["behaviour"])
```

```python
y_behav
```

```python
df
```

```python
from sklearn.model_selection import train_test_split

X=df.drop(["behaviour","circuit"],axis=1)

X_train, X_test, y_cir_train, y_cir_test, y_behav_train, y_behav_test = train_test_split(X, y_cir,y_behav, test_size=0.3, random_state=42)
```

```python
import joblib
import os

# Create artifacts directory
os.makedirs("artifacts", exist_ok=True)

# Save feature datasets
joblib.dump(X_train, "artifacts/X_train.pkl")
joblib.dump(X_test, "artifacts/X_test.pkl")

# Save circuit labels
joblib.dump(y_cir_train, "artifacts/y_cir_train.pkl")
joblib.dump(y_cir_test, "artifacts/y_cir_test.pkl")

# Save behaviour labels
joblib.dump(y_behav_train, "artifacts/y_behav_train.pkl")
joblib.dump(y_behav_test, "artifacts/y_behav_test.pkl")

# Save label encoders
joblib.dump(le_cir, "artifacts/le_cir.pkl")
joblib.dump(le_behav, "artifacts/le_behav.pkl")

print("All datasets and label encoders saved successfully.")
```
